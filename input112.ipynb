```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import pandas as pd
import time

# Setup WebDriver
options = webdriver.ChromeOptions()
driver = webdriver.Chrome(options=options)

# Buka halaman login
url = "http://112.140.160.228/pum/login"
driver.get(url)
time.sleep(2)  # Tunggu halaman termuat

# Login
driver.find_element(By.NAME, "USERNAME").send_keys("magangmsib")
driver.find_element(By.NAME, "PASSWORD").send_keys("msib")
driver.find_element(By.CLASS_NAME, "login100-form-btn").click()
time.sleep(3)  # Tunggu proses login

# Navigasi ke Data UMKM
driver.find_element(By.LINK_TEXT, "Data UMKM").click()
time.sleep(3)  # Tunggu halaman termuat

# Baca dataset Excel
file_path = "Dataset112.xlsx"
df = pd.read_excel(file_path)

df["Tanggal Event"] = pd.to_datetime(df["Tanggal Event"], errors="coerce").dt.strftime("%d-%m-%Y")

# List untuk menyimpan NIK yang tidak ditemukan
not_found_niks = []

# Looping untuk setiap baris data
for index, row in df.iterrows():
    nik = str(row["NIK"])  # Ambil NIK dari dataset

    # Cari NIK di search bar
    search_box = driver.find_element(By.CSS_SELECTOR, "input[type='search']")
    search_box.clear()
    search_box.send_keys(nik)
    search_box.send_keys(Keys.RETURN)
    time.sleep(3)  # Tunggu hasil pencarian

    # Cek apakah data ditemukan
    data_rows = driver.find_elements(By.XPATH, f"//td[contains(text(), '{nik}')]")
    if data_rows:
        print(f"Data ditemukan untuk NIK {nik}, masuk ke halaman edit...")

        # Klik tombol Edit
        edit_button = driver.find_element(By.XPATH, "//a[contains(@class, 'btn-warning') and contains(text(), 'Edit')]")
        edit_button.click()
        time.sleep(3)  # Tunggu halaman edit terbuka

        # Navigasi ke tab Intervensi
        intervensi_tab = driver.find_element(By.LINK_TEXT, "Intervensi")
        intervensi_tab.click()
        time.sleep(3)  # Tunggu halaman termuat

        # Isi formulir intervensi
        driver.find_element(By.NAME, "BULAN_TAHUN").send_keys(row["Tanggal Event"])
        driver.find_element(By.NAME, "intervensi").send_keys(row["Jenis Intervensi"])
        driver.find_element(By.NAME, "nama_kegiatan").send_keys(row["Nama Event"])
        driver.find_element(By.NAME, "OMSET").send_keys(str(row["Omset"]))

        # Klik tombol Tambah Data Intervensi
        driver.find_element(By.XPATH, "//button[contains(text(), 'Tambah Data Intervensi')]").click()
        time.sleep(3)  # Tunggu proses selesai

        print(f"Data intervensi untuk NIK {nik} berhasil ditambahkan.")

        # Kembali ke halaman Data UMKM
        driver.find_element(By.LINK_TEXT, "Data UMKM").click()
        time.sleep(3)  # Tunggu halaman termuat
    else:
        print(f"Data tidak ditemukan untuk NIK {nik}, lanjut ke NIK berikutnya...")
        not_found_niks.append(nik)  # Tambahkan ke daftar NIK yang tidak ditemukan

# Simpan daftar NIK yang tidak ditemukan ke file Excel
if not_found_niks:
    not_found_df = pd.DataFrame({'NIK Tidak Ditemukan': not_found_niks})
    not_found_df.to_excel("NIK_Tidak_Ditemukan.xlsx", index=False)
    print("Laporan NIK yang tidak ditemukan telah disimpan ke 'NIK_Tidak_Ditemukan.xlsx'")

# Tutup browser setelah selesai
driver.quit()
print("Proses selesai!")
```
